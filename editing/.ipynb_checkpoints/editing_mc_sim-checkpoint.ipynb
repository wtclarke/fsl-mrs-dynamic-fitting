```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt

from fsl_mrs.utils import mrs_io
from fsl_mrs.utils import synthetic as syn
from fsl_mrs.utils.fitting import fit_FSLModel
from fsl_mrs.utils.misc import parse_metab_groups
from fsl_mrs.core import MRS

```

```python
basis_on = mrs_io.read_basis('basis_sets/uzay_svs_mpress_68_res_60_on/')
basis_off = mrs_io.read_basis('basis_sets/uzay_svs_mpress_68_res_60_off/')
basis_diff = mrs_io.read_basis('basis_sets/uzay_svs_mpress_68_res_60_diff/')

example_sep = mrs_io.read_FID('example_data/separate.nii.gz')
example_diff = mrs_io.read_FID('example_data/diff.nii.gz')
```

```python
from fsl_mrs.utils import dynamic as dyn
import pandas as pd

def run_iteration(lw, noise):

    syn_fid_on = syn.synthetic_from_basis.syntheticFromBasisFile(basis_on,broadening=lw,noisecovariance=[[noise]],concentrations={'GABA':3.5})
    syn_mrs_on = syn_fid_on[1] 
    syn_mrs_on.FID = syn_fid_on[0]
#     syn_mrs_on.plot()
    on_concs = syn_fid_on[2]

    syn_fid_off = syn.synthetic_from_basis.syntheticFromBasisFile(basis_off,broadening=lw,noisecovariance=[[noise]],concentrations=on_concs)
    syn_mrs_off = syn_fid_off[1] 
    syn_mrs_off.FID = syn_fid_off[0]
#     syn_mrs_off.plot()
#     plt.show()

    syn_fid_off_equiv = syn.synthetic_from_basis.syntheticFromBasisFile(basis_off,broadening=lw,noisecovariance=[[noise/1.4]],concentrations=on_concs)
    syn_mrs_off_equiv = syn_fid_off_equiv[1] 
    syn_mrs_off_equiv.FID = syn_fid_off_equiv[0]

#     print(np.allclose(on_concs,syn_fid_off[2]))

    syn_fid_diff = syn_fid_on[0] - syn_fid_off[0] * (syn_mrs_on.scaling['basis']/syn_mrs_off.scaling['basis'])
    syn_mrs_diff = MRS(FID=syn_fid_diff,cf=syn_mrs_on.centralFrequency, bw=syn_mrs_on.bandwidth, basis=basis_diff)
    syn_mrs_diff.keep = ['NAA', 'NAAG', 'GABA', 'GSH','Glu', 'Gln']
    syn_mrs_diff.processForFitting()
#     syn_mrs_diff.plot()
#     plt.show()

    # Fit portion
    mg = [0,] * len(on_concs)
    fitargs = {'model':'lorentzian', 'baseline_order':0, 'metab_groups': mg}
    res_on = fit_FSLModel(syn_mrs_on, **fitargs,)
    res_off = fit_FSLModel(syn_mrs_off, **fitargs)
    res_off_equiv = fit_FSLModel(syn_mrs_off_equiv, **fitargs)
    fitargs_diff = {'model':'lorentzian', 'baseline_order':0, 'metab_groups': [0,] * 6}
    res_diff = fit_FSLModel(syn_mrs_diff, **fitargs_diff)

    df_res = pd.concat(
    (res_off.getConc(function=None),
     res_off_equiv.getConc(function=None),
     res_on.getConc(function=None),
     res_diff.getConc(function=None) * (syn_mrs_diff.scaling['basis']/syn_mrs_on.scaling['basis']) * (syn_mrs_on.scaling['FID']/syn_mrs_diff.scaling['FID'])
     ))
    df_res.set_index(np.array(['Off','Off Matched','On','Diff']), inplace=True)

    mrs_list = [syn_mrs_off, syn_mrs_on]
    tvar = [0, 1]
    dyn_obj = dyn.dynMRS(mrs_list, tvar, 'editconfig.py',rescale=False, **fitargs)
    init = dyn_obj.initialise()
    dyn_res = dyn_obj.fit(init=init)
    
    df_dyn = dyn_res['result'].data_frame.filter(like='conc', axis=1)
    df_dyn.columns = df_res.columns[:-1]
    df_dyn.set_index(np.array(['Dyn']), inplace=True)

    return pd.concat((df_res, df_dyn)), {'dyn': dyn_res, 'diff':(syn_mrs_diff, res_diff),'off':(syn_mrs_off_equiv, res_off_equiv)}
```

```python
df_container = []
reps = 500
# Scan time decreases by three times in each step
noise_vec = [np.round(25*(np.sqrt(3)**x)) for x in range(7)]
for noise in noise_vec:
    df_tmp = []
    for idx in range(reps):
        df_out,_ = run_iteration((3*6,0), noise)
        df_tmp.append(df_out)
    df_container.append(df_tmp)
```

```python
concs = pd.Series(np.asarray(on_concs),index=basis_off.names)
concs['Cr+PCr'] = concs['Cr'] + concs['PCr']
concs['Glx'] = concs['Glu'] + concs['Gln']
concs['Glx+GSH'] = concs['Glu'] + concs['Gln'] + concs['GSH']
concs['NAA+NAAG'] = concs['NAA'] + concs['NAAG']
concs
```

```python
df_all = tmp = []
for dfc in df_container:
    tmp.append(pd.concat(dfc, axis=0, keys=np.arange(0,len(dfc))))
df_all = pd.concat(tmp,keys=noise_vec)
df_all.index.rename(('Noise', 'MC rep', 'Fit Type'),inplace=True)
df_all.columns.rename('Metabolite',inplace=True)
df_all['Cr+PCr'] = df_all['Cr'] + df_all['PCr']
df_all['NAA+NAAG'] = df_all['NAA'] + df_all['NAAG']
df_all['Glx'] = df_all['Glu'] + df_all['Gln']
df_all['Glx+GSH'] = df_all['Glu'] + df_all['Gln'] + df_all['GSH']

# df_all.mean(level=[0,2])
df_all.std(level=[0,2])
```

```python
import seaborn as sns
df_all_diff = 100*(df_all - concs)/concs
df_all_diff.columns.rename('Metabolite',inplace=True)
df_all_diff = df_all_diff.filter(['NAA+NAAG', 'GABA', 'Glx']).drop(labels=['On', 'Off'],level=2)

df_plot2 = df_all_diff.stack().reset_index().rename(columns={0:'Values'})
# df_plot2

# fig = plt.figure(figsize=(15,8))
# sns.barplot(x='Metabolite', y='Values', data=df_plot2, hue='Fit Type')
g = sns.catplot(
    x='Noise',
    y='Values',
    data=df_plot2,
    hue='Fit Type',
    col='Metabolite', col_wrap=3,
    kind='bar',
    sharey=False)
plt.show()
```

```python
df_plot = df_all.filter(['NAA+NAAG', 'GABA', 'Glx']).drop(labels=['On', 'Off'],level=2)
df_plot = df_plot.stack().reset_index().rename(columns={0:'Values'})
df_plot

# fig = plt.figure(figsize=(15,8))
g = sns.catplot(
    x='Noise',
    y='Values',
    data=df_plot,
    hue='Fit Type',
    col='Metabolite', col_wrap=4,
    kind='bar',
    sharey=False)

for ax, col_names in zip(g.axes, g.col_names):
    conc = concs[col_names]
    ax.set_ylim([conc-conc*0.1, conc+conc*0.1])
    ax.hlines(conc, -1, 9, 'k', linestyles=':')
    ax.set_xlim([-0.5, 7.5])

plt.show()
```

```python
# RMSE
df_RMSE = (df_all - concs).pow(2).mean(axis=0,level=[0,2]).pow(0.5)
df_RMSE.columns.rename('Metabolite',inplace=True)

df_RMSE_reduced = df_RMSE.filter(['NAA+NAAG', 'Cr+PCr', 'GABA', 'Glx']).drop(labels=['On', 'Off'],level=1)
df_RMSE_reduced.rename(index={'Off Matched':'OFF', 'Diff':'DIFF', 'Dyn':'DYN'},inplace=True)
df_RMSE_to_plot = df_RMSE_reduced/df_RMSE_reduced.xs('OFF',level=1)
# df_RMSE_to_plot
df_RMSE_to_plot = df_RMSE_to_plot.stack().reset_index().rename(columns={0:'RMSE'})
# df_RMSE_to_plot

g = sns.barplot(data=df_RMSE_to_plot,x='Metabolite',y='RMSE',hue='Fit Type',ci='sd')
```
